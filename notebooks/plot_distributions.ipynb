```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

INSTANCE_SAMPLE_CSV = "/mnt/d/Downloads/Senior_Thesis/INSTANCE/Instance_sample_dataset_v3/metadata/metadata_Instance_events_10k.csv"
UNIFIED_CSV = "/home/aidar/study/senior_thesis/data/unified/metadata_events.csv"

print("Loading CSVs...")
instance_df = pd.read_csv(INSTANCE_SAMPLE_CSV, low_memory=False)
unified_df  = pd.read_csv(UNIFIED_CSV)

instance_mags = instance_df["source_magnitude"].dropna()
unified_mags  = unified_df["source_magnitude"].dropna()

print(f"INSTANCE sample : {len(instance_mags):,} events")
print(f"Unified dataset : {len(unified_mags):,} events")

bins = np.arange(
    min(instance_mags.min(), unified_mags.min()) - 0.25,
    max(instance_mags.max(), unified_mags.max()) + 0.5,
    0.25
)

fig, ax = plt.subplots(figsize=(14, 5), dpi=200)

# ── Left: overlaid normalized histograms ─────────────────────────────────────
ax.hist(
    unified_mags, bins=bins,
    density=False, alpha=0.6,
    color=(0.85, 0.35, 0.0), edgecolor='white', linewidth=0.3,
    label=(
        f'Unified dataset  (n={len(unified_mags):,})'
        f'\nmean={unified_mags.mean():.2f}'
        f'\nmedian={unified_mags.median():.2f}'
        f'\nstd={unified_mags.std():.2f}\n'
        )
)
ax.hist(
    instance_mags, bins=bins,
    density=False, alpha=0.6,
    color=(0.0, 0.35, 0.85), edgecolor='white', linewidth=0.3,
    label=(
        f'INSTANCE sample  (n={len(instance_mags):,})'
        f'\nmean={instance_mags.mean():.2f}'
        f'\nmedian={instance_mags.median():.2f}'
        f'\nstd={instance_mags.std():.2f}\n'
        )   
    )
ax.set_xlabel('Magnitude')
ax.set_ylabel('Counts')
ax.set_title('Magnitude Distribution', fontweight='bold', fontsize=13, y=1.0)
ax.legend(fontsize=9)
ax.set_xlim(bins[0], bins[-1])
ax.set_ylim(bottom=0.0)

# Bin boundary markers for the three magnitude classes
ax.axvline(3.0, color='black', linestyle='--', linewidth=0.8, alpha=1)
ax.axvline(5.0, color='black', linestyle='--', linewidth=0.8, alpha=1)

plt.tight_layout()

out_path = "magnitude_distribution_comparison.jpg"
plt.savefig(out_path, dpi=200, bbox_inches='tight')
plt.close(fig)

print(f"\nPlot saved to `{out_path}`")

# ── Per-bin breakdown ─────────────────────────────────────────────────────────
print("\nPer-bin breakdown:")
print(f"{'Bin':<20} {'INSTANCE sample':>16} {'Unified':>10} {'Unified %':>10}")
print("-" * 58)
for label, lo, hi in [("low  [1.0, 3.0)", 1.0, 3.0), ("mid  [3.0, 5.0)", 3.0, 5.0), ("high [5.0,  ∞ )", 5.0, 99)]:
    i_n = ((instance_mags >= lo) & (instance_mags < hi)).sum()
    u_n = ((unified_mags  >= lo) & (unified_mags  < hi)).sum()
    print(f"{label:<20} {i_n:>16,} {u_n:>10,} {u_n/len(unified_mags)*100:>9.1f}%")
```
